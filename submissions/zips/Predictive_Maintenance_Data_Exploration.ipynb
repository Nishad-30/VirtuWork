## Load Datasets

```
# Example placeholders – update paths with real datasets
# sensor_df = pd.read_csv('sensor_telemetry.csv')
# maintenance_df = pd.read_csv('maintenance_logs.csv')
# operations_df = pd.read_csv('operational_parameters.csv')

# For demonstration, we create a small synthetic dataset
np.random.seed(42)
sensor_df = pd.DataFrame({
    'machine_id': np.repeat([1,2,3], 100),
    'temperature': np.random.normal(70, 5, 300),
    'vibration': np.random.normal(0.5, 0.1, 300),
    'pressure': np.random.normal(30, 3, 300),
})

sensor_df.head()
```

## Data Exploration

```
sensor_df.describe()
```

## Missing Values Handling

```
sensor_df.isnull().sum()
```

## Outlier Treatment

```
for col in ['temperature', 'vibration', 'pressure']:
    q1 = sensor_df[col].quantile(0.25)
    q3 = sensor_df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    sensor_df[col] = sensor_df[col].clip(lower, upper)

sensor_df.describe()
```

## Feature Scaling

```
scaler = StandardScaler()
scaled_features = scaler.fit_transform(sensor_df[['temperature','vibration','pressure']])
scaled_df = pd.DataFrame(scaled_features, columns=['temperature','vibration','pressure'])
scaled_df.head()
```
